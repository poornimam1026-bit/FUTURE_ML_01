```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/train.csv')

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df.head()
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Day"] = df["Order Date"].dt.day
df["DayOfWeek"] = df["Order Date"].dt.dayofweek
```

```python
daily_sales = df.groupby("Order Date")["Sales"].sum().reset_index()
```

```python
daily_sales.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(daily_sales["Order Date"], daily_sales["Sales"])

plt.title("Historical Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.show()
```

```python
daily_sales["Time"] = np.arange(len(daily_sales))
```

```python
X = daily_sales[["Time"]]

y = daily_sales["Sales"]
```

```python
split_index = int(len(daily_sales) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred[:10])
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_sales["Order Date"].iloc[split_index:],
    y_test,
    label="Actual Sales"
)

plt.plot(
    daily_sales["Order Date"].iloc[split_index:],
    y_pred,
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Sales")

plt.xlabel("Date")
plt.ylabel("Sales")

plt.legend()

plt.show()
```

```python
future_days = 30

future_time = np.arange(
    len(daily_sales),
    len(daily_sales) + future_days
).reshape(-1, 1)

future_predictions = model.predict(future_time)

last_date = daily_sales["Order Date"].max()

future_dates = pd.date_range(
    start=last_date + pd.Timedelta(days=1),
    periods=future_days
)

forecast_df = pd.DataFrame({
    "Date": future_dates,
    "Predicted Sales": future_predictions
})

forecast_df
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_sales["Order Date"],
    daily_sales["Sales"],
    label="Historical Sales"
)

plt.plot(
    forecast_df["Date"],
    forecast_df["Predicted Sales"],
    label="Future Forecast"
)

plt.title("Sales Forecast for Next 30 Days")

plt.xlabel("Date")
plt.ylabel("Sales")

plt.legend()

plt.show()
```
